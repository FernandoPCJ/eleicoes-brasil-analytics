```python
import pandas as pd
from pathlib import Path
```

```python
arquivo = Path("../data/processed/candidatos_2026.parquet")

df = pd.read_parquet(arquivo)

df.head()
```

```python
df.shape
```

```python
dim_candidato = df[
    [
        "SQ_CANDIDATO",
        "nome_candidato",
        "nome_urna",
        "idade",
        "genero",
        "cor_raca",
        "ocupacao",
        "escolaridade",
        "estado_civil"
    ]
].copy()

dim_candidato.head()
```

```python
dim_candidato.shape
```

```python
dim_partido = df[
    [
        "NR_PARTIDO",
        "partido",
        "nome_partido",
        "NR_FEDERACAO",
        "NM_FEDERACAO"
    ]
].drop_duplicates()

dim_partido.head()
```

```python
dim_partido.shape
```

```python
dim_cargo = df[
    [
        "CD_CARGO",
        "cargo",
        "tipo_cargo"
    ]
].drop_duplicates()

dim_cargo.head()
```

```python
dim_cargo.shape
```

```python
dim_estado = df[
    [
        "estado",
        "estado_nome"
    ]
].drop_duplicates()

dim_estado.head()
```

```python
dim_estado.shape
```

```python
dim_estado.sort_values("estado")
```

```python
dim_estado = dim_estado[
    dim_estado["estado"] != "BR"
]

dim_estado.shape
```

```python
dim_estado = (
    dim_estado
    .sort_values("estado")
    .reset_index(drop=True)
)

dim_estado["id_estado"] = dim_estado.index + 1

dim_estado = dim_estado[
    [
        "id_estado",
        "estado",
        "estado_nome"
    ]
]

dim_estado.head()
```

```python
fato_candidatura = df[
    [
        "SQ_CANDIDATO",
        "NR_PARTIDO",
        "CD_CARGO",
        "estado",
        "ANO_ELEICAO",
        "situacao_final"
    ]
].copy()

fato_candidatura.head()
```

```python
fato_candidatura.shape
```

```python
fato_candidatura = fato_candidatura.merge(
    dim_estado,
    on="estado",
    how="left"
)

fato_candidatura.head()
```

```python
fato_candidatura["id_estado"].isna().sum()
```

```python
fato_candidatura[
    fato_candidatura["id_estado"].isna()
][
    [
        "estado",
        "SQ_CANDIDATO",
        "CD_CARGO"
    ]
]
```

```python
fato_candidatura = fato_candidatura[
    fato_candidatura["id_estado"].notna()
]
```

```python
fato_candidatura = fato_candidatura.drop(
    columns=[
        "estado",
        "estado_nome"
    ]
)

fato_candidatura.head()
```

```python
dim_partido = (
    dim_partido
    .sort_values("NR_PARTIDO")
    .reset_index(drop=True)
)

dim_partido["id_partido"] = dim_partido.index + 1


dim_partido = dim_partido[
    [
        "id_partido",
        "NR_PARTIDO",
        "partido",
        "nome_partido",
        "NR_FEDERACAO",
        "NM_FEDERACAO"
    ]
]

dim_partido.head()
```

```python
fato_candidatura = fato_candidatura.drop(
    columns=["id_partido_x"]
)

fato_candidatura = fato_candidatura.rename(
    columns={
        "id_partido_y": "id_partido"
    }
)
```

```python
dim_cargo = (
    dim_cargo
    .sort_values("CD_CARGO")
    .reset_index(drop=True)
)

dim_cargo["id_cargo"] = dim_cargo.index + 1


dim_cargo = dim_cargo[
    [
        "id_cargo",
        "CD_CARGO",
        "cargo",
        "tipo_cargo"
    ]
]


dim_cargo.head()
```
