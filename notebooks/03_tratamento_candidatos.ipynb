```python
import pandas as pd
from pathlib import Path


BASE_DIR = Path.cwd()

if BASE_DIR.name == "notebooks":
    BASE_DIR = BASE_DIR.parent


RAW_DATA = BASE_DIR / "data" / "raw" / "consulta_cand_2026_BRASIL.csv"

PROCESSED_DATA = BASE_DIR / "data" / "processed"


print(RAW_DATA)
print(PROCESSED_DATA)
```

```python
df = pd.read_csv(
    RAW_DATA,
    sep=";",
    encoding="latin1"
)

df.shape
```

```python
df_tratado = df.copy()

df_tratado.head()
```

```python
colunas_remover = [
    "NR_CPF_CANDIDATO",
    "NR_TITULO_ELEITORAL_CANDIDATO",
    "DS_EMAIL"
]


df_tratado = df_tratado.drop(
    columns=colunas_remover
)


df_tratado.shape
```

```python
renomear = {
    "NM_CANDIDATO": "nome_candidato",
    "NM_URNA_CANDIDATO": "nome_urna",
    "DS_CARGO": "cargo",
    "SG_PARTIDO": "partido",
    "NM_PARTIDO": "nome_partido",
    "SG_UF": "estado",
    "NM_UE": "estado_nome",
    "DS_GENERO": "genero",
    "DS_GRAU_INSTRUCAO": "escolaridade",
    "DS_ESTADO_CIVIL": "estado_civil",
    "DS_COR_RACA": "cor_raca",
    "DS_OCUPACAO": "ocupacao",
    "DS_SIT_TOT_TURNO": "situacao_final"
}


df_tratado = df_tratado.rename(
    columns=renomear
)


df_tratado.head()
```

```python
df_tratado.info()
```

```python
valores_invalidos = [
    "#NULO",
    "#NE"
]


df_tratado = df_tratado.replace(
    valores_invalidos,
    "Não informado"
)


df_tratado.head()
```

```python
df_tratado["DT_NASCIMENTO"] = pd.to_datetime(
    df_tratado["DT_NASCIMENTO"],
    errors="coerce",
    dayfirst=True
)


data_referencia = pd.Timestamp("2026-10-04")


df_tratado["idade"] = (
    (data_referencia - df_tratado["DT_NASCIMENTO"])
    .dt.days
    // 365
)


df_tratado[["nome_candidato","idade"]].head()
```

```python
def classificar_cargo(cargo):

    executivo = [
        "PRESIDENTE",
        "GOVERNADOR",
        "VICE-GOVERNADOR",
        "VICE-PRESIDENTE"
    ]

    legislativo = [
        "SENADOR",
        "DEPUTADO FEDERAL",
        "DEPUTADO ESTADUAL",
        "DEPUTADO DISTRITAL"
    ]

    if cargo in executivo:
        return "Executivo"

    elif cargo in legislativo:
        return "Legislativo"

    else:
        return "Outro"


df_tratado["tipo_cargo"] = (
    df_tratado["cargo"]
    .apply(classificar_cargo)
)


df_tratado["tipo_cargo"].value_counts()
```

```python
df_tratado[
[
"nome_candidato",
"cargo",
"partido",
"estado",
"idade",
"tipo_cargo"
]
].head(10)
```

```python
df_tratado.to_parquet(
    PROCESSED_DATA / "candidatos_2026.parquet"
)
```
